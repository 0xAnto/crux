Each converter downloads its raw data and writes JSONL files to `data/processed/<source>/<setting>/<split>.jsonl`.
RuleTaker and ProofWriter use depth 3 (questions of depth 0 to 3) for train and dev: `ruletaker/depth3/` and `proofwriter/OWA-depth3/`. Step 4 keeps ProofWriter depth 5 for the test only. BoardgameQA writes `boardgameqa/<subset>/`. DynaBench and Super-NaturalInstructions write `dynabench/` and `supernat/`.

```
!python -m crux.data.ruletaker
!python -m crux.data.proofwriter
!python -m crux.data.boardgameqa
!python -m crux.data.dynabench
!python -m crux.data.supernat
```
